```python
f = open('24 (3).txt').readline()

def sogl(x):
    s = 0
    for i in x:
        if i in 'BCDF':
            s+=1
    return s
def glas(x):
    s = 0
    for i in x:
        if i in 'AE':
            s+=1
    return s
    
a = [0 for x in range(len(f))]

for i in range(len(f)-4):
    if sogl(f[i:i+5])>glas(f[i:i+5]):
        if i<5:
            a[i] = 1
        else:
            a[i] = 1 + a[i-5]
            
print(max(a))
```
