```python
#1
f = open('26_9.txt').readlines()[1:]

a = []
for s in f:
    p = int(s.split()[0])
    l = str(s.split()[1])
    a.append([p, l])

a = sorted(a)
k = 0
# со срезами будь максимально аккуратен, потому что срез a[:n] берет первые n элементов
# поэтому sum(a[:n]) тоже считает сумму первых n элементов
# Почему? потому что у нас нумерация с 0, а элемент с индексом n не включаем
# У тебя эта ошибка во всём коде была
for c in range(1, 10**10):
    if sum(x for x, y in a[:c])<=50000:
        k = c
    # забыл условие для break, а то цикл до 10**10 слишком длинный
    if sum(x for x, y in a[:c])>50000:
        break
    
# print(k)
des = []
wor = []
for x in a:
    c = int(x[0])
    lit = str(x[1])
    if lit == 'G':
        des.append(c)
    else:
        wor.append(c)

# обязатльно отсортировать, чтобы брать минимальные зарплаты
des = sorted(des)
wor = sorted(wor)

#нам нужно максимизировать n, поэтому идем с конца и 1ое которое встретим, оно и будет максимальным
for n in range(2035, 0, -1): 
    su = sum(des[:n])+sum(wor[:k-n])
    if su <= 50000:
        print(n) #максимальное число дизайнеров
        print(50000-su) #остаток
        break
```

```python

with open("26_12.txt", "r") as f:
    lines = f.read().strip().split()
n = int(lines[0])
times = [int(lines[i]) for i in range(1, n + 1)]
times.sort()

registered = 0
ignored_streak = 0
max_ignored_streak = 0

group_start_time = -10000
count_in_group = 0

for t in times:
    if t - group_start_time >= 1000:
        group_start_time = t
        count_in_group = 1
        registered += 1
        ignored_streak = 0
    else:
        if count_in_group < 5:
            count_in_group += 1
            registered += 1
            ignored_streak = 0
        else:
            ignored_streak += 1
            if ignored_streak > max_ignored_streak:
                max_ignored_streak = ignored_streak

print(registered, max_ignored_streak)
```
