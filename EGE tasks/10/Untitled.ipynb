```python
from re import*
f = open('10_1.txt').readlines()
a = ' '.join(f)
a = sub(r'\b[а-яА-Я]\b', '', a)

k = 0
for x in a:
    if x in 'АЕЁИОУЫЭЮЯаеёиоуыэюя':
        k+=1

print(k)
```

```python
from re import*
f = open('10_4322.txt', encoding='utf-8').readlines()
f = ''.join(f)
r = r'(?:\w+-?)+'
c=0
f = findall(r,f)
for el in f:
    if el.count('в')+el.count('В')+el.count('B')>=1 and el.count('а')+el.count('А')+el.count('о')+el.count('О')==0:
        c+=1
print(c)
```
