```python
f = open('26_18621.txt').readlines()
n = int(f[0])
u = []
for s in f[1:]:
    a = [int(x) for x in s.split()]
    u.append(a)

def pl(x):
    su = 0
    for i in x[1:]:
        if i>0: su+=i
    return su

def ot(x):
    count = 0
    for i in x[1:]:
        if i!=0: count+=1
    return count

u = sorted(u, key = lambda x: [-sum(x[1:]), -pl(x), -ot(x), x[0]])

st = u[n//4]
st = [sum(st[1:]), pl(st), ot(st)]
idf = 0
for x in u[n//4+1:]:
    if [sum(x[1:]), pl(x), ot(x)] != st:
        idf = x[0]
        break

s = u[1699]
s = [sum(s[1:]), pl(s), ot(s)]
k = 0
for x in u:
    if [sum(x[1:]), pl(x), ot(x)] == s:
        k+=1
        
print(idf, k)
```
