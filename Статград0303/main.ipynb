```python
def f(n):
    a = 2
    s = []
    k=0
    while a**2<=n:
        if n%a==0:
            k+=1
            n//=a
            if '2' in str(a) or '3' in str(a): 
                s.append(a)
                if len(s)>3: return False
        else:
            a+=1
    if n>1:
        k+=1
        if '2' in str(n) or '3' in str(n): 
            s.append(n)
    if len(s) == 3 and k==3: return True
    else: return False
k = 0
for x in range(5000000, 10**10):
    if f(x) and k<5:
        print(x)
        k+=1
    if k==5:
        break
```
