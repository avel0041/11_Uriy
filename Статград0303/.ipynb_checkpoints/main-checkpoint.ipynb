```python
#3 78914
f = open('Статград2701/26 (2).txt').readlines()[1:]
time = [0]*15*60
st = []
k = 0

for x in f[:102]:
    k+=1
    st.append([int(x), k])

for s in f[103:]:
    m, prod = [int(x) for x in s.split()]
    if m<15*60:
        if m+prod<15*60:
            for i in range(m, m+prod+1):
                time[i]+=1
        else:
            for i in range(m, 15*60):
                time[i]+=1
    else: continue
print(time)
```
