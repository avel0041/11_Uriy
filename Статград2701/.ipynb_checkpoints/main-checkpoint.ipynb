```python
from functools import*

@lru_cache(None)
def g(n):
    if  n>=384_242: return n/4 + 18
    else: return 12 + g(n+41)
for n in range(10**6, -1, -1):
    g(n)
def f(n):
    if  n>=20: return f(n-4) + 4620
    else: return 8*(g(n-12)-21)
for n in range(914):
    f(n)
print(f(913))
#2703082
```

```python
f = open('17.txt').readlines()
a = [int(x) for x in f]
ma = max([x for x in a if x%100 == 70])
k = []

for i in range(len(a)-2):
    # c1, c2, c3 = a[i:i+3]
    if (a[i]>=0 and a[i+1]>=0 and a[i+2]>=0) and (a[i]+a[i+1]+a[i+2]<=ma):
        k.append(a[i]+a[i+1]+a[i+2])
print(len(k), max(k))
```
