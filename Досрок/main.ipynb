```python
#25

def f(x):
    for i in range(2, int(x**0.5)+1):
        if x%i == 0:
            if i%10 == 7 and i!=7:
                return True
            if (x//i)%10 == 7 and (x//i)!=7:
                return True
    return False

def g(x):
    a = set()
    for i in range(2, int(x**0.5)+1):
        if x%i == 0:
            if i%10 == 7 and i!=7:
                a.add(i)
            if (x//i)%10 == 7 and (x//i)!=7:
                a.add(x//i)
    return min(a)

k = 0
for x in range(700_001, 10**10):
    if k>=5: break
    if f(x):
        k+=1
        print(x, g(x))

# 700002 27
# 700003 37
# 700005 6087
# 700007 77
# 700008 29167
```

```python
#26

f = open('26.txt').readlines()
n = int(f[0])

sr = 0
data = []
for s in f[1:]:
    art, pr, st = [int(x) for x in s.split()]
    sr+=pr
    data.append([art, pr, st])
sr = sr/n

dor = []
for x in data:
    if x[1]>sr:
        dor.append(x)
dor = sorted(dor)
nach = dor[0][0]
kone = dor[-1][0]

s = [[x, 0, 0, 0, 0] for x in range(nach, kone+1)]
for x in dor:
    for y in s:
        if y[0] == x[0]:
            y[1] = x[1]
            if x[2] == 0:
                y[2]+=1
                y[4]+=x[1]
            if x[2] == 1:
                y[3]+=1
s = sorted(s, key = lambda x: [x[2], x[1], -x[3]])
print(s[-1][4], s[-1][3])

# 43656 36
```

```python
#27

from math import*
f = open('27_B.txt').readlines()

data = []
for s in f:
    x, y, st = [x for x in s.split()]
    x = float(x)
    y = float(y)
    data.append([x, y, st[0], int(st[1]), st[2:]])

clusters = []
while data:
    cl = [data.pop()]
    for p in cl:
        points = [x for x in data if dist(p[:2], x[:2])<1]
        for x in points:
            cl.append(x)
            data.remove(x)
    clusters.append(cl)


# from turtle import*
# from random import*
# tracer(0)
# pu()
# screensize(2000, 2000)
# k = 30

# for cl in clusters:
#     color = random(), random(), random()
#     for x, y, d, e, r in cl:
#         goto(x*k, y*k)
#         dot(7, color)
# done()

def cen(cl):
    a = []
    for p in cl:
        a+=[(sum(dist(p[:2], x[:2]) for x in cl), p)]
    return min(a)[1]

y = []
for cl in clusters:
    for p in cl:
        if p[2] == 'C' and p[4] == 'XXL':
            for x in cl:
                if x[2] == p[2] and x[4] == p[4] and x!=p:
                    y.append(dist(p[:2], x[:2]))
b1 = int(min(y)*100_000)

dop = []
for cl in clusters:
    kol = 0
    for p in cl:
        if p[2] == 'R' and p[4] == 'XT':
            kol+=1
    dop.append([cl, kol])

dop = sorted(dop, key = lambda x: [x[1], x[0]])
c1 = cen(dop[0][0])
c2 = cen(dop[-1][0])
b2 = int(dist(c1[:2], c2[:2])*100_000)

print(b1, b2)

# 589 797325
```
