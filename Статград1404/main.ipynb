```python
#27a
from math import*
f = open('27_A.txt').readlines()

data = []
for s in f:
    x, y, asd = [x for x in s.replace(',','.').split()]
    x = float(x)
    y = float(y)
    spec = asd[0]
    spepc = asd[1]
    svet = asd[2:]
    data.append([x, y, spec, spepc, svet])

clusters = []
while data:
    cl = [data.pop()]
    for p in cl:
        points = [x for x in data if dist(p[:2], x[:2]) < 1]
        for x in points:
            cl.append(x)
            data.remove(x)
    clusters.append(cl)

from turtle import*
from random import*

tracer(0)
screensize(2000, 2000)
pu()
k = 15

for cl in clusters:
    color = random(), random(), random()
    for x, y, a, s, d in cl:
        goto(x*k, y*k)
        dot(5, color)
done()

def cen(cl):
    a = []
    for p in cl:
        a+=[(sum(dist(p[:2], x[:2]) for x in cl), p)]
    return min(a)[1]

clusters = sorted(clusters, key = len)
cmin = cen(clusters[0])
a1sp = []
for x in clusters[0]:
    if x[2] == 'A' and x[4] == 'III':
        a1sp.append([dist(x[:2], cmin[:2]), x[0], x[1]])
a1 = min(a1sp)[1]
a2 = min(a1sp)[2]
print(int(a1*10_000), int(a2*10_000))

# 390205 182699
```
